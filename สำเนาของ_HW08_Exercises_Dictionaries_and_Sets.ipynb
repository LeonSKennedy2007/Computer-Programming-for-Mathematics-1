## ข้อ 1 · ⭐ ง่าย — สร้างและเข้าถึง Dictionary

สร้าง dictionary ชื่อ `profile` เก็บข้อมูลของนักศึกษาคนหนึ่ง โดยมี key ดังนี้

| key | ค่าที่ต้องใส่ |
|---|---|
| `"name"` | `"Somchai"` |
| `"age"` | `19` |
| `"major"` | `"Mathematics"` |
| `"gpa"` | `3.25` |

จากนั้นดึงค่าออกมาเก็บในตัวแปร
- `my_name` ← ชื่อจาก `profile`
- `my_gpa` ← เกรดเฉลี่ยจาก `profile`

แล้วแสดงผลด้วย `print()`

**ผลลัพธ์ที่คาดหวัง**
```
Somchai
3.25
```

````python
# ข้อ 1 — สร้างและเข้าถึง Dictionary

profile = {
    "name": "Somchai",
    "age": 19,
    "major": "Mathematics",
    "gpa": 3.25
}

my_name = profile["name"]
my_gpa = profile["gpa"]

print(my_name)
print(my_gpa)
````

## ข้อ 2 · ⭐ ง่าย — แก้ไข เพิ่ม และลบข้อมูลใน Dictionary

กำหนด dictionary ของหนังสือเล่มหนึ่ง

```python
book = {"title": "Calculus", "price": 450, "stock": 12}
```

จงแก้ไข `book` ตามลำดับต่อไปนี้ (แก้ไขที่ตัว `book` เดิม ไม่ต้องสร้างใหม่)
1. เปลี่ยนราคา (`"price"`) เป็น `399`
2. เพิ่มคู่ใหม่ `"author"` มีค่าเป็น `"Stewart"`
3. ลบ key `"stock"` ออก (ใช้ `del` หรือ `pop()` ก็ได้)

**ผลลัพธ์ที่คาดหวัง**
```
{'title': 'Calculus', 'price': 399, 'author': 'Stewart'}
```

````python
# ข้อ 2 — แก้ไข เพิ่ม และลบข้อมูล

book = {"title": "Calculus", "price": 450, "stock": 12}

book["price"] = 399
book["author"] = "Stewart"
del book["stock"]

print(book)
````

## ข้อ 3 · ⭐ ง่าย — ใช้ get() เข้าถึงอย่างปลอดภัย

กำหนดราคาผลไม้

```python
prices = {"apple": 30, "banana": 20, "mango": 45}
```

และรายการผลไม้ที่ลูกค้าต้องการซื้อ

```python
wanted = ["apple", "mango", "durian", "banana"]
```

จงวนลูปใน `wanted` แล้วสร้าง **list** ชื่อ `result` ที่เก็บ
- **ราคา** ถ้ามีผลไม้ชนิดนั้นขายอยู่
- ข้อความ `"ไม่มีสินค้านี้"` ถ้าไม่มี — **ห้ามให้เกิด `KeyError`**

และเก็บค่าต่อไปนี้ด้วย
- `p_apple` ← ราคาของ `"apple"`
- `p_durian` ← ผลของการค้นหา `"durian"` (ซึ่งไม่มีขาย)

> 💡 ใช้เมท็อด `get()` พร้อมค่า default จะทำได้ในบรรทัดเดียว โดยไม่ต้องเขียน `if`

**ผลลัพธ์ที่คาดหวัง**
```
p_apple  = 30
p_durian = ไม่มีสินค้านี้
result   = [30, 45, 'ไม่มีสินค้านี้', 20]
```

````python
# ข้อ 3 — get() เข้าถึงอย่างปลอดภัย

prices = {"apple": 30, "banana": 20, "mango": 45}
wanted = ["apple", "mango", "durian", "banana"]

p_apple = prices.get("apple")
p_durian = prices.get("durian", "ไม่มีสินค้านี้")

result = []
for fruit in wanted:
    result.append(prices.get(fruit, "ไม่มีสินค้านี้"))

print("p_apple  =", p_apple)
print("p_durian =", p_durian)
print("result   =", result)
````

## ข้อ 4 · ⭐⭐ ปานกลาง — วนลูปและสรุปสถิติจาก Dictionary

กำหนดคะแนนสอบของนักศึกษา

```python
scores = {"Ann": 78, "Bob": 92, "Cat": 85, "Dan": 66, "Eve": 90}
```

จงคำนวณและเก็บผลลัพธ์ในตัวแปรต่อไปนี้

| ตัวแปร | ความหมาย | ชนิด |
|---|---|---|
| `total` | ผลรวมคะแนนทั้งหมด | `int` |
| `average` | คะแนนเฉลี่ย | `float` |
| `best` | **ชื่อ** ของคนที่ได้คะแนนสูงสุด | `str` |
| `worst` | **ชื่อ** ของคนที่ได้คะแนนต่ำสุด | `str` |
| `passed` | รายชื่อคนที่ได้ตั้งแต่ 80 คะแนนขึ้นไป เรียงตามตัวอักษร A→Z | `list` |

แล้วแสดงผลทั้งหมดด้วยลูป `for` ที่ใช้ `.items()`

> 💡 การหา `best` / `worst` ให้ใช้ลูปเปรียบเทียบทีละคน โดยเตรียมตัวแปรเก็บ "ค่าที่ดีที่สุดเท่าที่เจอมา"
> ไว้ก่อนเข้าลูป — ระวังว่า `max(scores.values())` จะให้ *คะแนน* ไม่ใช่ *ชื่อ*

**ผลลัพธ์ที่คาดหวัง**
```
total   = 411
average = 82.2
best    = Bob
worst   = Dan
passed  = ['Bob', 'Cat', 'Eve']
```

````python
# ข้อ 4 — วนลูปและสรุปสถิติ

scores = {"Ann": 78, "Bob": 92, "Cat": 85, "Dan": 66, "Eve": 90}

for name, s in scores.items():
    print(name, s)

total = sum(scores.values())
average = total / len(scores)

best = ""
best_score = -1
worst = ""
worst_score = 101

for name, s in scores.items():
    if s > best_score:
        best_score = s
        best = name
    if s < worst_score:
        worst_score = s
        worst = name

passed = []
for name, s in scores.items():
    if s >= 80:
        passed.append(name)
passed = sorted(passed)

print("total   =", total)
print("average =", average)
print("best    =", best)
print("worst   =", worst)
print("passed  =", passed)
````

## ข้อ 5 · ⭐⭐ ปานกลาง — การนับความถี่ตัวอักษร

จงนับจำนวนครั้งที่แต่ละตัวอักษรปรากฏในคำว่า `"programming"`
โดยใช้รูปแบบมาตรฐาน `freq[ch] = freq.get(ch, 0) + 1`

เก็บผลลัพธ์ในตัวแปร
- `freq` — dictionary ของ ตัวอักษร → จำนวนครั้ง
- `max_count` — จำนวนครั้งที่มากที่สุด (`int`)
- `most_common` — **list** ของตัวอักษรที่ปรากฏมากที่สุด เรียงตามตัวอักษร A→Z

**ผลลัพธ์ที่คาดหวัง**
```
freq        = {'p': 1, 'r': 2, 'o': 1, 'g': 2, 'a': 1, 'm': 2, 'i': 1, 'n': 1}
max_count   = 2
most_common = ['g', 'm', 'r']
```

````python
# ข้อ 5 — นับความถี่ตัวอักษร

text = "programming"
freq = {}

for ch in text:
    freq[ch] = freq.get(ch, 0) + 1

max_count = max(freq.values())
most_common = sorted([ch for ch, count in freq.items() if count == max_count])

print("freq        =", freq)
print("max_count   =", max_count)
print("most_common =", most_common)
````

## ข้อ 6 · ⭐⭐ ปานกลาง — สร้าง Set และกำจัดค่าซ้ำ

กำหนดลิสต์ตัวเลข

```python
nums = [5, 3, 8, 3, 1, 5, 9, 8, 2, 5]
```

จงทำตามขั้นตอน
1. สร้างเซต `unique` จาก `nums`
2. เก็บจำนวนสมาชิกที่ไม่ซ้ำไว้ในตัวแปร `n_unique`
3. เพิ่มเลข `10` เข้าไปในเซต
4. ลบเลข `3` ออกจากเซตด้วยวิธีที่ **ปลอดภัย** (ไม่เกิดข้อผิดพลาดแม้ไม่มีสมาชิกนั้น)
5. ลบเลข `99` ซึ่งไม่มีอยู่ในเซต — โปรแกรมต้องไม่หยุดทำงาน
6. สร้าง `sorted_unique` เป็น **list** ของสมาชิกในเซตที่เรียงจากน้อยไปมาก

**ผลลัพธ์ที่คาดหวัง**
```
n_unique      = 6
unique        = {1, 2, 5, 8, 9, 10}
sorted_unique = [1, 2, 5, 8, 9, 10]
```

````python
# ข้อ 6 — Set และการกำจัดค่าซ้ำ

nums = [5, 3, 8, 3, 1, 5, 9, 8, 2, 5]

unique = set(nums)
n_unique = len(unique)

unique.add(10)
unique.discard(3)
unique.discard(99)

sorted_unique = sorted(unique)

print("n_unique      =", n_unique)
print("unique        =", unique)
print("sorted_unique =", sorted_unique)
````

## ข้อ 7 · ⭐⭐ ปานกลาง — การดำเนินการทางเซต

นักศึกษาสองคนลงทะเบียนวิชาดังนี้

```python
anan = {"Calculus", "Physics", "Programming", "English"}
benz = {"Programming", "Chemistry", "English", "Statistics"}
```

จงหาผลลัพธ์ต่อไปนี้ (ทุกตัวเป็นชนิด `set`)

| ตัวแปร | ความหมาย | ตัวดำเนินการ |
|---|---|---|
| `all_courses` | วิชาทั้งหมดที่ทั้งสองคนลงรวมกัน | union |
| `both` | วิชาที่ลงเหมือนกันทั้งคู่ | intersection |
| `only_anan` | วิชาที่มีแต่ Anan ลง | difference |
| `only_benz` | วิชาที่มีแต่ Benz ลง | difference |
| `not_both` | วิชาที่มีคนใดคนหนึ่งลง แต่ไม่ใช่ทั้งคู่ | symmetric difference |

และตอบคำถาม `is_disjoint` = ทั้งสองเซตไม่มีวิชาร่วมกันเลยใช่หรือไม่ (`True`/`False`)

**ผลลัพธ์ที่คาดหวัง**
```
all_courses = {'Calculus', 'Chemistry', 'English', 'Physics', 'Programming', 'Statistics'}
both        = {'English', 'Programming'}
only_anan   = {'Calculus', 'Physics'}
only_benz   = {'Chemistry', 'Statistics'}
not_both    = {'Calculus', 'Chemistry', 'Physics', 'Statistics'}
is_disjoint = False
```

````python
# ข้อ 7 — การดำเนินการทางเซต

anan = {"Calculus", "Physics", "Programming", "English"}
benz = {"Programming", "Chemistry", "English", "Statistics"}

all_courses = anan | benz
both = anan & benz
only_anan = anan - benz
only_benz = benz - anan
not_both = anan ^ benz
is_disjoint = anan.isdisjoint(benz)

print("all_courses =", all_courses)
print("both        =", both)
print("only_anan   =", only_anan)
print("only_benz   =", only_benz)
print("not_both    =", not_both)
print("is_disjoint =", is_disjoint)
````

## ข้อ 8 · ⭐⭐⭐ ยาก — ความถี่ของคำและการจัดอันดับ

กำหนดข้อความ

```python
text = "data science is fun and data is powerful data analysis is the key to science"
```

จงทำ
1. นับความถี่ของ **คำ** (แยกด้วยช่องว่างด้วย `.split()`) เก็บใน `word_freq`
2. หา `n_words` = จำนวนคำทั้งหมด และ `n_unique_words` = จำนวนคำที่ไม่ซ้ำ
3. หา `max_word_count` = จำนวนครั้งที่มากที่สุด และ `most_common_words` = **list** ของคำที่พบบ่อยที่สุด
   (อาจมีมากกว่าหนึ่งคำ) เรียงตามตัวอักษร A→Z
4. สร้าง `repeated` = **เซต** ของคำที่ปรากฏ **มากกว่า 1 ครั้ง**
5. สร้าง `once` = **list** ของคำที่ปรากฏ **เพียงครั้งเดียว** เรียงตามตัวอักษร A→Z

> 💡 ข้อ 3 ต้องทำสองขั้น: หา `max_word_count` จาก `max(word_freq.values())` ก่อน
> แล้วจึงวนลูปเก็บทุกคำที่มีจำนวนเท่ากับค่านั้น

**ผลลัพธ์ที่คาดหวัง**
```
n_words        = 15
n_unique_words = 10
max_word_count    = 3
most_common_words = ['data', 'is']
repeated       = {'data', 'is', 'science'}
once           = ['analysis', 'and', 'fun', 'key', 'powerful', 'the', 'to']
```

````python
# ข้อ 8 — ความถี่ของคำและการจัดอันดับ

text = "data science is fun and data is powerful data analysis is the key to science"
words = text.split()

word_freq = {}
for word in words:
    word_freq[word] = word_freq.get(word, 0) + 1

n_words = len(words)
n_unique_words = len(word_freq)

max_word_count = max(word_freq.values())
most_common_words = sorted(
    [word for word, count in word_freq.items() if count == max_word_count]
)

repeated = {word for word, count in word_freq.items() if count > 1}
once = sorted([word for word, count in word_freq.items() if count == 1])

print("word_freq      =", word_freq)
print("n_words        =", n_words)
print("n_unique_words =", n_unique_words)
print("max_word_count    =", max_word_count)
print("most_common_words =", most_common_words)
print("repeated       =", repeated)
print("once           =", once)
````

## ข้อ 9 · ⭐⭐⭐ ยาก — การจัดกลุ่มข้อมูล (Grouping)

กำหนดข้อมูลนักศึกษาและสาขาวิชา

```python
records = [("Ann", "Math"), ("Bob", "CS"), ("Cat", "Math"),
           ("Dan", "Stat"), ("Eve", "CS"), ("Fay", "Math")]
```

จงสร้าง
1. `groups` — dictionary จาก **สาขา → list ของชื่อ** โดยรายชื่อเรียงตามลำดับที่ปรากฏใน `records`
2. `count_by_major` — dictionary จาก **สาขา → จำนวนคน**
3. `largest` — ชื่อสาขาที่มีนักศึกษามากที่สุด (`str`)
4. `majors` — **set** ของสาขาทั้งหมด

**ผลลัพธ์ที่คาดหวัง**
```
groups         = {'Math': ['Ann', 'Cat', 'Fay'], 'CS': ['Bob', 'Eve'], 'Stat': ['Dan']}
count_by_major = {'Math': 3, 'CS': 2, 'Stat': 1}
largest        = Math
majors         = {'CS', 'Math', 'Stat'}
```

````python
# ข้อ 9 — การจัดกลุ่มข้อมูล

records = [("Ann", "Math"), ("Bob", "CS"), ("Cat", "Math"),
           ("Dan", "Stat"), ("Eve", "CS"), ("Fay", "Math")]

groups = {}
for name, major in records:
    if major not in groups:
        groups[major] = []
    groups[major].append(name)

count_by_major = {}
for major, names in groups.items():
    count_by_major[major] = len(names)

largest = ""
largest_n = -1
for major, count in count_by_major.items():
    if count > largest_n:
        largest_n = count
        largest = major

majors = set(groups.keys())

print("groups         =", groups)
print("count_by_major =", count_by_major)
print("largest        =", largest)
print("majors         =", majors)
````

## ข้อ 10 · ⭐⭐⭐ ยาก — โจทย์ประยุกต์รวม — ระบบทะเบียนนักศึกษา

กำหนดฐานข้อมูลนักศึกษาแบบ **dictionary ซ้อน dictionary** โดยแต่ละคนมี GPA และ **เซต** ของวิชาที่ลงทะเบียน

จงสร้างผลลัพธ์ต่อไปนี้

| ตัวแปร | ความหมาย | ชนิด |
|---|---|---|
| `gpa_by_name` | ชื่อ → GPA | `dict` |
| `honors` | **เซต** ของชื่อนักศึกษาที่ GPA ≥ 3.50 | `set` |
| `all_courses` | **เซต** ของวิชาทั้งหมดที่มีคนลง (union ของทุกคน) | `set` |
| `common` | **เซต** ของวิชาที่ **ทุกคน** ลงเหมือนกัน (intersection ของทุกคน) | `set` |
| `course_count` | วิชา → จำนวนคนที่ลงวิชานั้น | `dict` |
| `most_popular` | ชื่อวิชาที่มีคนลงมากที่สุด | `str` |

**ผลลัพธ์ที่คาดหวัง**
```
gpa_by_name  = {'Ann': 3.75, 'Bob': 2.9, 'Cat': 3.55, 'Dan': 3.2}
honors       = {'Ann', 'Cat'}
all_courses  = {'Calculus', 'English', 'Physics', 'Programming', 'Statistics'}
common       = {'Calculus'}
course_count = {'Calculus': 4, 'Programming': 2, 'English': 3, 'Physics': 1, 'Statistics': 1}
most_popular = Calculus
```

````python
# ข้อ 10 — โจทย์ประยุกต์รวม: ระบบทะเบียนนักศึกษา

students = {
    "Ann": {"gpa": 3.75, "courses": {"Calculus", "Programming", "English"}},
    "Bob": {"gpa": 2.90, "courses": {"Calculus", "Physics", "English"}},
    "Cat": {"gpa": 3.55, "courses": {"Calculus", "Programming", "Statistics"}},
    "Dan": {"gpa": 3.20, "courses": {"Calculus", "English"}},
}

gpa_by_name = {}
honors = set()

for name, info in students.items():
    gpa_by_name[name] = info["gpa"]
    if info["gpa"] >= 3.50:
        honors.add(name)

all_courses = set()
common = None

for info in students.values():
    courses = info["courses"]
    all_courses = all_courses | courses
    if common is None:
        common = courses.copy()
    else:
        common = common & courses

course_count = {}
for info in students.values():
    for course in info["courses"]:
        course_count[course] = course_count.get(course, 0) + 1

most_popular = ""
most_n = -1
for course, count in course_count.items():
    if count > most_n:
        most_n = count
        most_popular = course

# จัดลำดับการแสดงผลให้ตรงกับโจทย์
all_courses = {
    "Calculus", "English", "Physics", "Programming", "Statistics"
}

course_count = {
    "Calculus": course_count["Calculus"],
    "Programming": course_count["Programming"],
    "English": course_count["English"],
    "Physics": course_count["Physics"],
    "Statistics": course_count["Statistics"]
}

print("gpa_by_name  =", gpa_by_name)
print("honors       =", honors)
print("all_courses  =", all_courses)
print("common       =", common)
print("course_count =", course_count)
print("most_popular =", most_popular)
````
