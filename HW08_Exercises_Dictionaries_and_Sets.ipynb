กำหนดฐานข้อมูลนักศึกษาแบบ **dictionary ซ้อน dictionary** โดยแต่ละคนมี GPA และ **เซต** ของวิชาที่ลงทะเบียน

```python
students = {
    "Ann": {"gpa": 3.75, "courses": {"Calculus", "Programming", "English"}},
    "Bob": {"gpa": 2.90, "courses": {"Calculus", "Physics", "English"}},
    "Cat": {"gpa": 3.55, "courses": {"Calculus", "Programming", "Statistics"}},
    "Dan": {"gpa": 3.20, "courses": {"Calculus", "English"}},
}
```

จงสร้างผลลัพธ์ต่อไปนี้

| ตัวแปร | ความหมาย | ชนิด |
|---|---|---|
| `gpa_by_name` | ชื่อ → GPA | `dict` |
| `honors` | **เซต** ของชื่อนักศึกษาที่ GPA ≥ 3.50 | `set` |
| `all_courses` | **เซต** ของวิชาทั้งหมดที่มีคนลง (union ของทุกคน) | `set` |
| `common` | **เซต** ของวิชาที่ **ทุกคน** ลงเหมือนกัน (intersection ของทุกคน) | `set` |
| `course_count` | วิชา → จำนวนคนที่ลงวิชานั้น | `dict` |
| `most_popular` | ชื่อวิชาที่มีคนลงมากที่สุด | `str` |

**ผลลัพธ์ที่คาดหวัง**
```
gpa_by_name  = {'Ann': 3.75, 'Bob': 2.9, 'Cat': 3.55, 'Dan': 3.2}
honors       = {'Ann', 'Cat'}
all_courses  = {'Calculus', 'English', 'Physics', 'Programming', 'Statistics'}
common       = {'Calculus'}
course_count = {'Calculus': 4, 'Programming': 2, 'English': 3, 'Physics': 1, 'Statistics': 1}
most_popular = Calculus
```
