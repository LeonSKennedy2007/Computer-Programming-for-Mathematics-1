## ข้อ 2 · การตัดแบ่งลิสต์ (Slicing)

**สิ่งที่ต้องทำ** — ใช้ slicing เท่านั้น (ห้ามใช้ลูป) เพื่อแสดง

1. สมาชิก 3 ตัวแรก
2. สมาชิก 3 ตัวสุดท้าย
3. สมาชิกตั้งแต่ดัชนี 2 ถึงดัชนี 6
4. สมาชิกแบบเว้นตำแหน่ง (ดัชนี 0, 2, 4, ...)
5. ลิสต์ที่เรียงกลับด้าน

**ผลลัพธ์ที่ต้องได้**

```text
3 ตัวแรก: [5, 10, 15]
3 ตัวสุดท้าย: [40, 45, 50]
ดัชนี 2 ถึง 6: [15, 20, 25, 30, 35]
เว้นตำแหน่ง: [5, 15, 25, 35, 45]
กลับด้าน: [50, 45, 40, 35, 30, 25, 20, 15, 10, 5]
```


````python
nums = [5, 10, 15, 20, 25, 30, 35, 40, 45, 50]

# ===== เขียนโค้ดของคุณที่นี่ =====

print("3 ตัวแรก:", nums[:3])
print("3 ตัวสุดท้าย:", nums[-3:])
print("ดัชนี 2 ถึง 6:", nums[2:7])
print("เว้นตำแหน่ง:", nums[::2])
print("กลับด้าน:", nums[::-1])
````

## ข้อ 3 · เพิ่มสมาชิกด้วย append() และ insert()

**สิ่งที่ต้องทำ**

1. เพิ่ม `"mango"` ต่อท้ายลิสต์
2. แทรก `"orange"` ที่ตำแหน่งดัชนี 1
3. แสดงลิสต์ผลลัพธ์และจำนวนสมาชิก

**ผลลัพธ์ที่ต้องได้**

```text
ลิสต์เริ่มต้น: ['apple', 'banana', 'cherry']
หลังเพิ่มและแทรก: ['apple', 'orange', 'banana', 'cherry', 'mango']
```

## ข้อ 4 · ลบสมาชิกด้วย remove() และ pop()

**สิ่งที่ต้องทำ** — ทำตามลำดับและแสดงผลทุกขั้นตอน

1. ลบค่า `25` ตัวแรกที่พบ ด้วย `remove()`
2. ดึงสมาชิกตำแหน่ง 0 ออกด้วย `pop()` และแสดงค่าที่ถูกดึงออก
3. ดึงสมาชิกตัวสุดท้ายออกด้วย `pop()` และแสดงค่าที่ถูกดึงออก
4. แสดงลิสต์สุดท้าย

**ผลลัพธ์ที่ต้องได้**

```text
ลิสต์เริ่มต้น: [10, 25, 30, 25, 40, 55]
หลัง remove(25): [10, 30, 25, 40, 55]
ค่าที่ pop ตำแหน่ง 0: 10
ค่าที่ pop ตัวสุดท้าย: 55
ลิสต์สุดท้าย: [30, 25, 40]
```


````python
nums = [10, 25, 30, 25, 40, 55]

# ===== เขียนโค้ดของคุณที่นี่ =====

print("ลิสต์เริ่มต้น:", nums)

nums.remove(25)
print("หลัง remove(25):", nums)

value = nums.pop(0)
print("ค่าที่ pop ตำแหน่ง 0:", value)

value = nums.pop()
print("ค่าที่ pop ตัวสุดท้าย:", value)

print("ลิสต์สุดท้าย:", nums)
````

## ข้อ 5 · sort() ต่างจาก sorted() อย่างไร

**สิ่งที่ต้องทำ**

1. สร้างลิสต์ใหม่ที่เรียงจากน้อยไปมากด้วย `sorted()` แล้วแสดงผล
2. แสดงลิสต์เดิมซ้ำ เพื่อยืนยันว่า `sorted()` ไม่เปลี่ยนลิสต์เดิม
3. เรียงลิสต์เดิมจากมากไปน้อยด้วย `sort(reverse=True)` แล้วแสดงผล

**ผลลัพธ์ที่ต้องได้**

```text
ลิสต์เดิม: [45, 12, 78, 33, 56, 12, 90]
เรียงน้อยไปมาก (ลิสต์ใหม่): [12, 12, 33, 45, 56, 78, 90]
ลิสต์เดิมหลังใช้ sorted(): [45, 12, 78, 33, 56, 12, 90]
หลัง sort(reverse=True): [90, 78, 56, 45, 33, 12, 12]
```


````python
data = [45, 12, 78, 33, 56, 12, 90]

# ===== เขียนโค้ดของคุณที่นี่ =====

print("ลิสต์เดิม:", data)

new_data = sorted(data)
print("เรียงน้อยไปมาก (ลิสต์ใหม่):", new_data)

print("ลิสต์เดิมหลังใช้ sorted():", data)

data.sort(reverse=True)
print("หลัง sort(reverse=True):", data)
````

## ข้อ 6 · ผลรวมและค่าเฉลี่ยด้วยลูป for

**ข้อมูลที่กำหนดให้:** อุณหภูมิสูงสุดรายวัน 7 วัน

**สิ่งที่ต้องทำ**

1. หาผลรวมและค่าเฉลี่ย โดย **ห้ามใช้ `sum()`** (ต้องใช้ลูป `for` กับตัวแปรสะสม)
2. นับจำนวนวันที่อุณหภูมิสูงกว่าค่าเฉลี่ย
3. แสดงผลรวมและค่าเฉลี่ยเป็นทศนิยม 2 ตำแหน่ง

**ผลลัพธ์ที่ต้องได้**

```text
ผลรวม = 212.00
ค่าเฉลี่ย = 30.29
จำนวนวันที่สูงกว่าค่าเฉลี่ย = 3 วัน
```


````python
temps = [28.5, 31.0, 29.5, 33.5, 30.0, 27.5, 32.0]

# ===== เขียนโค้ดของคุณที่นี่ =====


total = 0

for temp in temps:
    total += temp

average = total / len(temps)

count = 0
for temp in temps:
    if temp > average:
        count += 1

print(f"ผลรวม = {total:.2f}")
print(f"ค่าเฉลี่ย = {average:.2f}")
print(f"จำนวนวันที่สูงกว่าค่าเฉลี่ย = {count} วัน")
````

## ข้อ 7 · หาค่ามากสุด–น้อยสุด พร้อมตำแหน่ง

**สิ่งที่ต้องทำ** — หาค่ามากที่สุดและน้อยที่สุด พร้อมบอกตำแหน่ง (ดัชนี) ของค่านั้น

> **ห้ามใช้** `max()`, `min()`, `sort()`, `index()`
>
> คำใบ้: กำหนดค่าเริ่มต้นเป็นสมาชิกตัวแรก แล้ววนด้วย `for i in range(len(nums)):` เพื่อเปรียบเทียบทีละตัว
> ถ้ามีค่าซ้ำกัน ให้ตอบตำแหน่งที่พบครั้งแรก

**ผลลัพธ์ที่ต้องได้**

```text
ค่ามากที่สุด = 91 อยู่ที่ตำแหน่ง 2
ค่าน้อยที่สุด = 8 อยู่ที่ตำแหน่ง 1
```


````python
nums = [23, 8, 91, 45, 8, 67, 34]

# ===== เขียนโค้ดของคุณที่นี่ =====

maximum = nums[0]
minimum = nums[0]
max_pos = 0
min_pos = 0

for i in range(len(nums)):
    if nums[i] > maximum:
        maximum = nums[i]
        max_pos = i

    if nums[i] < minimum:
        minimum = nums[i]
        min_pos = i

print(f"ค่ามากที่สุด = {maximum} อยู่ที่ตำแหน่ง {max_pos}")
print(f"ค่าน้อยที่สุด = {minimum} อยู่ที่ตำแหน่ง {min_pos}")
````

## ข้อ 8 · การกรองข้อมูล (Filtering) — แยกเลขคู่และเลขคี่

**สิ่งที่ต้องทำ** — วนลูปผ่าน `nums` แล้ว

1. เก็บเลขคู่ลง `evens` และเลขคี่ลง `odds` ด้วย `append()`
2. แสดงลิสต์ทั้งสองพร้อมจำนวนสมาชิก
3. แสดงผลรวมของเลขคู่

**ผลลัพธ์ที่ต้องได้**

```text
เลขคู่: [12, 8, 20, 14, 16] จำนวน 5 ตัว
เลขคี่: [7, 5, 3, 9] จำนวน 4 ตัว
ผลรวมของเลขคู่ = 70
```


````python
nums = [12, 7, 5, 8, 20, 3, 14, 9, 16]

evens = []
odds = []

# ===== เขียนโค้ดของคุณที่นี่ =====

for num in nums:
    if num % 2 == 0:
        evens.append(num)
    else:
        odds.append(num)

even_sum = 0
for num in evens:
    even_sum += num

print(f"เลขคู่: {evens} จำนวน {len(evens)} ตัว")
print(f"เลขคี่: {odds} จำนวน {len(odds)} ตัว")
print(f"ผลรวมของเลขคู่ = {even_sum}")
````

## ข้อ 9 · การแปลงข้อมูล (Transformation) — พื้นที่วงกลม

**ข้อมูลที่กำหนดให้:** รัศมีของวงกลม 5 วง

**สิ่งที่ต้องทำ**

1. สร้างลิสต์ `areas` เก็บพื้นที่วงกลมของแต่ละรัศมี ($A = \pi r^2$) ปัดทศนิยม 2 ตำแหน่งด้วย `round(x, 2)`
2. แสดงลิสต์ `areas`
3. แสดงผลรวมพื้นที่ทั้งหมด (ทศนิยม 2 ตำแหน่ง)

**ผลลัพธ์ที่ต้องได้**

```text
พื้นที่วงกลม: [3.14, 12.57, 28.27, 50.27, 78.54]
ผลรวมพื้นที่ = 172.79
```


````python
import math

radii = [1, 2, 3, 4, 5]
areas = []

# ===== เขียนโค้ดของคุณที่นี่ =====

radii = [1, 2, 3, 4, 5]
areas = []

for r in radii:
    area = math.pi * r ** 2
    areas.append(round(area, 2))

print("พื้นที่วงกลม:", areas)
print(f"ผลรวมพื้นที่ = {sum(areas):.2f}")
````

## ข้อ 10 · ตรวจสมาชิกด้วย in / not in — ป้องกันข้อมูลซ้ำ

**ข้อมูลที่กำหนดให้:** `students` คือรหัสที่ลงทะเบียนแล้ว · `new_ids` คือรหัสที่ขอเพิ่ม

**สิ่งที่ต้องทำ** — วนลูปผ่าน `new_ids` แล้ว

- ถ้ารหัสนั้นยังไม่มีใน `students` → เพิ่มเข้าไป และแจ้งว่าเพิ่มสำเร็จ
- ถ้ามีอยู่แล้ว → แจ้งว่าไม่เพิ่ม
- สุดท้ายแสดงรายชื่อทั้งหมดและจำนวนคน

**ผลลัพธ์ที่ต้องได้**

```text
เพิ่ม 6504 สำเร็จ
6502 มีอยู่แล้ว ไม่เพิ่ม
เพิ่ม 6505 สำเร็จ
6501 มีอยู่แล้ว ไม่เพิ่ม
เพิ่ม 6506 สำเร็จ
รายชื่อสุดท้าย: ['6501', '6502', '6503', '6504', '6505', '6506'] จำนวน 6 คน
```


````python
students = ["6501", "6502", "6503"]
new_ids = ["6504", "6502", "6505", "6501", "6506"]

# ===== เขียนโค้ดของคุณที่นี่ =====

for student_id in new_ids:
    if student_id not in students:
        students.append(student_id)
        print(f"เพิ่ม {student_id} สำเร็จ")
    else:
        print(f"{student_id} มีอยู่แล้ว ไม่เพิ่ม")

print(f"รายชื่อสุดท้าย: {students} จำนวน {len(students)} คน")
````

## ข้อ 11 · นับความถี่ด้วย count() และหาตำแหน่งด้วย index()

**ข้อมูลที่กำหนดให้:** `votes` คือผลการโหวต · `candidates` คือรายชื่อผู้สมัคร

**สิ่งที่ต้องทำ**

1. วนลูปผ่าน `candidates` แล้วแสดงคะแนนของแต่ละคน (ใช้ `count()`)
2. แสดงตำแหน่งแรกที่พบเสียงโหวตของ `"C"` (ใช้ `index()`)
3. หาและแสดงผู้ที่ได้คะแนนสูงสุด

**ผลลัพธ์ที่ต้องได้**

```text
ผู้สมัคร A ได้ 4 คะแนน
ผู้สมัคร B ได้ 2 คะแนน
ผู้สมัคร C ได้ 2 คะแนน
ผู้สมัคร D ได้ 1 คะแนน
ตำแหน่งแรกที่พบ C = 3
ผู้ชนะคือ A ด้วย 4 คะแนน
```


````python
votes = ["A", "B", "A", "C", "B", "A", "D", "A", "C"]
candidates = ["A", "B", "C", "D"]

# ===== เขียนโค้ดของคุณที่นี่ =====

for candidate in candidates:
    score = votes.count(candidate)
    print(f"ผู้สมัคร {candidate} ได้ {score} คะแนน")

print("ตำแหน่งแรกที่พบ C =", votes.index("C"))

winner = candidates[0]
highest_score = votes.count(winner)

for candidate in candidates:
    score = votes.count(candidate)
    if score > highest_score:
        highest_score = score
        winner = candidate

print(f"ผู้ชนะคือ {winner} ด้วย {highest_score} คะแนน")
````

```text
ค่าน้อยสุด: 8.0
เรียงมากไปน้อย: [25.0, 15.0, 12.0, 8.0]
```


````python
# (ข้อนี้ไม่มีข้อมูลตั้งต้น ให้รับค่าจากผู้ใช้เอง)
n = int(input("ต้องการป้อนกี่จำนวน? "))
# ===== เขียนโค้ดของคุณที่นี่ =====

numbers = []

for i in range(n):
    num = float(input(f"ตัวเลขที่ {i + 1}: "))
    numbers.append(num)

total = sum(numbers)
average = total / len(numbers)

descending = sorted(numbers, reverse=True)

print("ข้อมูลที่ป้อน:", numbers)
print("ผลรวม:", total)
print(f"ค่าเฉลี่ย: {average:.2f}")
print("ค่ามากสุด:", max(numbers))
print("ค่าน้อยสุด:", min(numbers))
print("เรียงมากไปน้อย:", descending)
````

## ข้อ 13 · ค่าเฉลี่ยและส่วนเบี่ยงเบนมาตรฐาน

**สูตรที่ใช้**

$$\bar{x} = \frac{\sum x_i}{n} \qquad \sigma^2 = \frac{\sum (x_i - \bar{x})^2}{n} \qquad \sigma = \sqrt{\sigma^2}$$

**สิ่งที่ต้องทำ** — แสดงผลเป็นทศนิยม 2 ตำแหน่ง

1. คำนวณค่าเฉลี่ยโดยใช้ `sum()` และ `len()`
2. คำนวณความแปรปรวนด้วยลูป `for` (สะสมค่า `(x - mean) ** 2`)
3. คำนวณส่วนเบี่ยงเบนมาตรฐานจาก `var ** 0.5`
4. นับจำนวนข้อมูลที่มากกว่าค่าเฉลี่ย

**ผลลัพธ์ที่ต้องได้**

```text
ค่าเฉลี่ย = 21.43
ความแปรปรวน = 38.82
ส่วนเบี่ยงเบนมาตรฐาน = 6.23
ข้อมูลที่มากกว่าค่าเฉลี่ย = 4 ค่า
```


````python
data = [12, 15, 18, 22, 25, 28, 30]

# ===== เขียนโค้ดของคุณที่นี่ =====

mean = sum(data) / len(data)

sum_squared = 0

for x in data:
    sum_squared += (x - mean) ** 2

var = sum_squared / len(data)
std = var ** 0.5

count = 0
for x in data:
    if x > mean:
        count += 1

print(f"ค่าเฉลี่ย = {mean:.2f}")
print(f"ความแปรปรวน = {var:.2f}")
print(f"ส่วนเบี่ยงเบนมาตรฐาน = {std:.2f}")
print(f"ข้อมูลที่มากกว่าค่าเฉลี่ย = {count} ค่า")
````

## ข้อ 14 · ลิสต์ซ้อนลิสต์ (Nested List) — เมทริกซ์ 3×3

**สิ่งที่ต้องทำ** — ใช้ลูปซ้อนลูป

1. แสดงเมทริกซ์ในรูปตาราง (แถวละบรรทัด)
2. แสดงผลรวมของแต่ละแถว
3. แสดงผลรวมของแต่ละหลัก (คอลัมน์)
4. แสดงผลรวมแนวทแยงหลัก คือ `matrix[0][0] + matrix[1][1] + matrix[2][2]`

**ผลลัพธ์ที่ต้องได้**

```text
3 7 2
8 1 9
4 6 5
ผลรวมแต่ละแถว: [12, 18, 15]
ผลรวมแต่ละคอลัมน์: [15, 14, 16]
ผลรวมแนวทแยง: 9
```


````python
matrix = [[3, 7, 2],
          [8, 1, 9],
          [4, 6, 5]]

# ===== เขียนโค้ดของคุณที่นี่ =====

row_sums = []
col_sums = []

# แสดงเมทริกซ์และหาผลรวมแต่ละแถว
for row in matrix:
    row_sum = 0

    for value in row:
        print(value, end=" ")
        row_sum += value

    row_sums.append(row_sum)
    print()

# หาผลรวมแต่ละคอลัมน์
for col in range(3):
    col_sum = 0

    for row in range(3):
        col_sum += matrix[row][col]

    col_sums.append(col_sum)

diagonal = 0

for i in range(3):
    diagonal += matrix[i][i]

print("ผลรวมแต่ละแถว:", row_sums)
print("ผลรวมแต่ละคอลัมน์:", col_sums)
print("ผลรวมแนวทแยง:", diagonal)
````

## ข้อ 15 · โจทย์ประยุกต์ — ระบบสรุปผลคะแนนสอบ

**ข้อมูลที่กำหนดให้:** `names` และ `scores` เป็นลิสต์คู่ขนาน (ดัชนีเดียวกันคือคนเดียวกัน)

**สิ่งที่ต้องทำ**

1. คำนวณและแสดงคะแนนเฉลี่ยของทั้งห้อง (ทศนิยม 2 ตำแหน่ง)
2. แสดงรายชื่อผู้ที่ได้คะแนนสูงกว่าค่าเฉลี่ย
3. แปลงคะแนนเป็นเกรดตามเกณฑ์ เก็บลงลิสต์ `grades` แล้วแสดงผลรายคน
4. นับจำนวนนักศึกษาในแต่ละเกรดด้วย `count()`

**เกณฑ์การตัดเกรด**

| คะแนน | เกรด |
| :-- | :-- |
| 80 ขึ้นไป | A |
| 70 – 79 | B |
| 60 – 69 | C |
| 50 – 59 | D |
| ต่ำกว่า 50 | F |

**ผลลัพธ์ที่ต้องได้**

```text
คะแนนเฉลี่ยของห้อง = 73.30
ผู้ที่ได้คะแนนสูงกว่าค่าเฉลี่ย: ['Ben', 'Cake', 'Earn', 'Ice', 'Kong']
Ann 55 -> D
Ben 78 -> B
Cake 92 -> A
Dao 43 -> F
Earn 88 -> A
Fah 67 -> C
Gun 71 -> B
Ice 95 -> A
Jib 60 -> C
Kong 84 -> A
สรุปจำนวนเกรด: A=4 B=2 C=2 D=1 F=1
```


````python
names = ["Ann", "Ben", "Cake", "Dao", "Earn", "Fah", "Gun", "Ice", "Jib", "Kong"]
scores = [55, 78, 92, 43, 88, 67, 71, 95, 60, 84]
grades = []

# ===== เขียนโค้ดของคุณที่นี่ =====


average = sum(scores) / len(scores)

print(f"คะแนนเฉลี่ยของห้อง = {average:.2f}")

above_average = []

for i in range(len(names)):
    if scores[i] > average:
        above_average.append(names[i])

print("ผู้ที่ได้คะแนนสูงกว่าค่าเฉลี่ย:", above_average)

grades = []

for score in scores:
    if score >= 80:
        grade = "A"
    elif score >= 70:
        grade = "B"
    elif score >= 60:
        grade = "C"
    elif score >= 50:
        grade = "D"
    else:
        grade = "F"

    grades.append(grade)

for i in range(len(names)):
    print(f"{names[i]} {scores[i]} -> {grades[i]}")

print(
    f"สรุปจำนวนเกรด: "
    f"A={grades.count('A')} "
    f"B={grades.count('B')} "
    f"C={grades.count('C')} "
    f"D={grades.count('D')} "
    f"F={grades.count('F')}"
)
````
