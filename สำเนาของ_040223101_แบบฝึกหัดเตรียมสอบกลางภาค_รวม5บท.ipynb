```python
# แสดงผลลัพธ์ของ 2 + 3 * 4
# ลำดับการคำนวณทางคอมพิวเตอร์และคณิตศาสตร์:
# 1. ทำการคูณ (*) ก่อน คือ 3 * 4 = 12
# 2. ทำการบวก (+) ทีหลัง คือ 2 + 12 = 14

result = 2 + 3 * 4
print(result)
```

**ข้อ 96.** เขียนโปรแกรมประมาณค่า sin(x) ด้วยอนุกรมเทย์เลอร์ x − x³/3! + x⁵/5! − ... (5 พจน์) เมื่อ x = 1 เรเดียน ทศนิยม 6 ตำแหน่ง

```python
# ✍️ เขียนโค้ดข้อ 96 ที่นี่
import math

x = 1
result = 0

for i in range(5):
    power = 2 * i + 1
    term = x ** power / math.factorial(power)

    if i % 2 == 0:
        result += term
    else:
        result -= term

print(f"{result:.6f}")
```

```python
# ✍️ เขียนโค้ดข้อ 97 ที่นี่
for i in range(7):
    for j in range(i + 1):
        numerator = 1
        denominator = 1

        for k in range(1, i + 1):
            numerator *= k

        for k in range(1, j + 1):
            denominator *= k

        for k in range(1, i - j + 1):
            denominator *= k

        print(numerator // denominator, end=" ")

    print()
```
